```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc


# 读取 CSV 文件
file_path = "../models_cla/results/ROC_val/classification_metrics_val.csv"  # 确保你的 CSV 文件在当前目录
df = pd.read_csv(file_path)

# **2. 提取 FPR（假正率）和 TPR（真正率）**
fpr = df["FPR"]
tpr = df["TPR"]

# **3. 计算 AUC（曲线下面积）**
roc_auc = auc(fpr, tpr)

# **4. 绘制 ROC 曲线**
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC Curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label="Random Guess")  # 随机分类的参考线
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("ROC Curve")
plt.legend(loc="lower right")
plt.grid()
plt.show()

# **5. 输出 AUC 值**
print(f"AUC Score: {roc_auc:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import auc, roc_curve

# **1. 读取 CSV 文件**
file_path = "../models_cla/results/ROC_test/classification_metrics_test.csv"  # 确保你的 CSV 文件路径正确
df = pd.read_csv(file_path)

# **2. 提取 FPR、TPR 和阈值**
fpr = df["FPR"][:33]
tpr = df["TPR"][:33]
thresholds = df["Residual"][:33]  # 假设 Residual 作为阈值

# **3. 排序保证 FPR 单调递增**
sorted_indices = np.argsort(fpr)
fpr = fpr.iloc[sorted_indices].values
tpr = tpr.iloc[sorted_indices].values
thresholds = thresholds.iloc[sorted_indices].values

# **4. 计算 Youden’s Index**
youden_index = tpr - fpr
best_threshold_youden = thresholds[np.argmax(youden_index)]  # 选择最大 Youden’s Index 的阈值

# **5. 计算距离 (0,1) 最近的点**
distances = np.sqrt((1 - tpr) ** 2 + fpr ** 2)
best_threshold_distance = thresholds[np.argmin(distances)]  # 选择距离 (0,1) 最近的阈值

# **6. 计算 AUC**
roc_auc = auc(fpr, tpr)

# **7. 绘制 ROC 曲线**
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC Curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label="Random Guess")

# **8. 标记最优阈值**
best_fpr_youden = fpr[np.argmax(youden_index)]
best_tpr_youden = tpr[np.argmax(youden_index)]
plt.scatter(best_fpr_youden, best_tpr_youden, marker='o', color='red', label=f'Best Threshold (Youden) = {best_threshold_youden:.2f}')

best_fpr_distance = fpr[np.argmin(distances)]
best_tpr_distance = tpr[np.argmin(distances)]
plt.scatter(best_fpr_distance, best_tpr_distance, marker='o', color='green', label=f'Best Threshold (Dist) = {best_threshold_distance:.2f}')

plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("ROC Curve with Best Threshold")
plt.legend(loc="lower right")
plt.grid()
plt.show()

# **9. 输出最优阈值**
print(f"Best Threshold (Youden’s Index): {best_threshold_youden:.2f}")
print(f"Best Threshold (Min Distance to (0,1)): {best_threshold_distance:.2f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 读取 CSV 文件
file_path = "../models_cla_v3/results/ROC_val_4/classification_metrics_val_4.csv"
# file_path = "../models_cla_v3/results/ROC_val_4/classification_metrics_val_4.csv"
df = pd.read_csv(file_path)

# 提取 Residual 作为横坐标，F1-Score 作为纵坐标
residuals = df["Tau"]
f1_scores = df["F1-Score"]

# 绘制 F1-Score 曲线
plt.figure(figsize=(10, 7))
plt.plot(residuals, f1_scores, marker='o', color='b', linestyle='-', label="F1-Score Curve", markersize=8)

# 找到 F1-Score 最高的点
max_index = np.argmax(f1_scores)
max_residual = residuals[max_index]
max_f1_score = f1_scores[max_index]

# 标记最高点
plt.scatter(max_residual, max_f1_score, color='red', s=100, label=f'Max F1-Score = {max_f1_score:.2f}', zorder=10)
# plt.text(max_residual, max_f1_score, f'({max_residual:.2f}, {max_f1_score:.2f})', 
#          fontsize=28, verticalalignment='bottom', horizontalalignment='right')

# 添加根引线（arrow）
plt.annotate(
    f'({max_residual:.2f}, {max_f1_score:.2f})', 
    xy=(max_residual, max_f1_score), 
    xytext=(max_residual + 1, max_f1_score - 0.05),  # 调整文字位置
    fontsize=20,
    arrowprops=dict(facecolor='black', arrowstyle="->", linewidth=2)
)

# 设置字体大小
plt.xlabel("Threshold", fontsize=20)
plt.ylabel("F1-Score", fontsize=20)
plt.title("F1-Score vs Threshold (Val set)", fontsize=20)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.legend(fontsize=20)
plt.grid(True)
plt.savefig("f1_score_vs_residual_test_tgt.png", transparent=True, bbox_inches='tight', dpi=300)

# 显示图像
plt.show()
```
