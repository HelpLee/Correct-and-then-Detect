# train_val_test_progression

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

# 总样本数
total_points = 15600
x = np.arange(total_points)

# 设置训练阶段样本点
train_sizes = [1560, 3120, 4680, 6240, 7800, 9360, 10920, 12480,]
val_size = 1560
test_size = 1560
val_start = total_points - (val_size + test_size)
test_start = total_points - test_size

# 创建动画帧数据
frames = []

for train_end in train_sizes:
    frame_labels = ['Train' if i < train_end else
                    'Val' if val_start <= i < test_start else
                    'Test' if i >= test_start else
                    'Unlabeled'
                    for i in range(total_points)]
    frames.append((frame_labels, train_end))

# 定义颜色映射
color_map = {'Train': 'blue', 'Val': 'green', 'Test': 'red', 'Unlabeled': 'lightgray'}

# 创建动画
fig, ax = plt.subplots(figsize=(14, 2.2))

def update(frame_data):
    frame_labels, train_end = frame_data
    ax.clear()
    ax.set_xlim(0, total_points)
    ax.set_ylim(0, 1.5)
    ax.set_yticks([])
    ax.set_xticks(np.linspace(0, total_points, 11, dtype=int))
    ax.set_title("Training Progression (Blue: Train, Green: Val, Red: Test)")

    # 绘制矩形块
    current = 0
    while current < total_points:
        label = frame_labels[current]
        start = current
        while current < total_points and frame_labels[current] == label:
            current += 1
        end = current
        ax.fill_betweenx([0, 1], start, end, color=color_map[label], linewidth=0)

    # 设置统一高度
    label_y = 1.35
    count_y = 1.12

    # 区域标签
    if train_end > 0:
        ax.text(train_end / 2, label_y, 'Train', ha='center', fontsize=11, color=color_map['Train'], weight='bold')
        ax.text(train_end / 2, count_y, f"{train_end} samples", ha='center', va='bottom',
                fontsize=10, color='blue', bbox=dict(facecolor='white', edgecolor='blue', boxstyle='round,pad=0.3'))

    ax.text((val_start + test_start) / 2, label_y, 'Validation', ha='center', fontsize=11, color=color_map['Val'], weight='bold')
    ax.text((val_start + test_start) / 2, count_y, f"{val_size} samples", ha='center', va='bottom',
            fontsize=10, color='green', bbox=dict(facecolor='white', edgecolor='green', boxstyle='round,pad=0.3'))

    ax.text((test_start + total_points) / 2, label_y, 'Test', ha='center', fontsize=11, color=color_map['Test'], weight='bold')
    ax.text((test_start + total_points) / 2, count_y, f"{test_size} samples", ha='center', va='bottom',
            fontsize=10, color='red', bbox=dict(facecolor='white', edgecolor='red', boxstyle='round,pad=0.3'))

ani = animation.FuncAnimation(fig, update, frames=frames, repeat=True, interval=800)

# 保存为 GIF
gif_path = "train_val_test_progression.gif"
ani.save(gif_path, writer='pillow', fps=1)

gif_path
```

```python
# Re-import required packages after kernel reset
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from scipy.stats.qmc import Halton
from mpl_toolkits.mplot3d import Axes3D

# 设置3D空间范围
x_range = (0, 1)
y_range = (0, 1)
z_range = (0, 1)

# 3D 网格采样
def generate_3d_grid_points(n_per_axis):
    x = np.linspace(x_range[0], x_range[1], n_per_axis)
    y = np.linspace(y_range[0], y_range[1], n_per_axis)
    z = np.linspace(z_range[0], z_range[1], n_per_axis)
    grid_x, grid_y, grid_z = np.meshgrid(x, y, z)
    return np.vstack([grid_x.ravel(), grid_y.ravel(), grid_z.ravel()]).T

# 3D Halton 采样
def generate_3d_halton_points(n_points):
    sampler = Halton(d=3, scramble=True)
    return sampler.random(n=n_points)

# 设置总点数
n_total = 125  # 5x5x5
n_per_axis = int(round(n_total ** (1/3)))
grid_points = generate_3d_grid_points(n_per_axis)
halton_points = generate_3d_halton_points(n_total)

# 初始化图形
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.set_xlim(x_range)
ax.set_ylim(y_range)
ax.set_zlim(z_range)
ax.set_title("3D Grid vs Halton Sampling")
grid_scat = ax.scatter([], [], [], c='blue', label='Grid')
halton_scat = ax.scatter([], [], [], c='red', marker='x', label='Halton')
ax.legend()

# 初始化
def init_3d():
    grid_scat._offsets3d = ([], [], [])
    halton_scat._offsets3d = ([], [], [])
    return grid_scat, halton_scat

# 更新
def update_3d(frame):
    n = frame + 1
    gx, gy, gz = grid_points[:n].T
    hx, hy, hz = halton_points[:n].T
    grid_scat._offsets3d = (gx, gy, gz)
    halton_scat._offsets3d = (hx, hy, hz)
    return grid_scat, halton_scat

# 动画生成
ani = animation.FuncAnimation(
    fig, update_3d, frames=n_total, init_func=init_3d, blit=False, interval=150, repeat=False
)

# 保存 GIF
gif_3d_path = "halton_vs_grid_3d.gif"
ani.save(gif_3d_path, writer='pillow')

gif_3d_path
```
