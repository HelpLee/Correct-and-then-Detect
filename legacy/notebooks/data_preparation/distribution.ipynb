```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
source_file = "../data/monitoring_logs_1000/merged_monitor_log.csv"
target_file = "../data/monitoring_logs_300/merged_monitor_log.csv"

source_data = pd.read_csv(source_file)
target_data = pd.read_csv(target_file)

# 提取 motor6_temperature 列
source_motor6_temp = source_data["motor6_temperature"]
target_motor6_temp = target_data["motor6_temperature"]

# 📌 **绘制直方图**
plt.figure(figsize=(10, 5))
plt.hist(source_motor6_temp, bins=range(int(source_motor6_temp.min()), int(source_motor6_temp.max()) + 1, 1), alpha=0.5, label="Source Data")
plt.hist(target_motor6_temp, bins=range(int(target_motor6_temp.min()), int(target_motor6_temp.max()) + 1, 1), alpha=0.5, label="Target Data")

# 设置标题和标签
plt.xlabel("Motor6 Temperature")
plt.ylabel("Frequency")
plt.title("Temperature Distribution of Source vs Target")
plt.legend()
plt.grid()
plt.show()


# 📌 **绘制折线图 - Source 数据**
plt.figure(figsize=(10, 5))
plt.plot(source_motor6_temp, label="Source Data", color='blue', alpha=0.7)
plt.xlabel("Time")
plt.ylabel("Motor6 Temperature")
plt.title("Source Data - Temperature Trend")
plt.legend()
plt.grid()
plt.show()


# 📌 **绘制折线图 - Target 数据**
plt.figure(figsize=(10, 5))
plt.plot(target_motor6_temp, label="Target Data", color='red', alpha=0.7)
plt.xlabel("Time")
plt.ylabel("Motor6 Temperature")
plt.title("Target Data - Temperature Trend")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ks_2samp
from sklearn.decomposition import PCA



data_file_source = '../data/source_all_data.csv'
source_df = pd.read_csv(data_file_source).dropna()
data_file_target = '../data/monitoring_logs_1000/merged_monitor_log.csv'
target_df = pd.read_csv(data_file_target).dropna()


numeric_cols = source_df.columns

# 2️⃣ 分布图（KDE + 直方图）
def plot_feature_distributions(source_df, target_df, features, bins=50):
    for col in features:
        plt.figure(figsize=(6, 3))
        sns.histplot(source_df[col], bins=bins, label="Source", color='skyblue', kde=True, stat='density')
        sns.histplot(target_df[col], bins=bins, label="Target", color='orange', kde=True, stat='density')
        plt.title(f"Feature Distribution: {col}")
        plt.legend()
        plt.tight_layout()
        plt.show()

plot_feature_distributions(source_df, target_df, numeric_cols)

# 3️⃣ 统计量对比
summary_df = pd.DataFrame({
    'source_mean': source_df.mean(),
    'target_mean': target_df.mean(),
    'source_std': source_df.std(),
    'target_std': target_df.std(),
    'mean_diff': source_df.mean() - target_df.mean(),
    'std_diff': source_df.std() - target_df.std()
}).round(3)

print("\n📊 Summary Statistics:")
print(summary_df)

# 4️⃣ KS 检验
print("\n📊 KS Test (p-value < 0.05 表示分布显著不同):")
for col in numeric_cols:
    stat, p = ks_2samp(source_df[col], target_df[col])
    mark = "✅ Diff" if p < 0.05 else "✅ Same"
    print(f"{col:16}: p = {p:.4f}  {mark}")

# 5️⃣ PCA 可视化
combined = pd.concat([source_df, target_df], ignore_index=True)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(combined)

plt.figure(figsize=(6, 4))
plt.scatter(X_pca[:len(source_df), 0], X_pca[:len(source_df), 1], alpha=0.5, label="Source")
plt.scatter(X_pca[len(source_df):, 0], X_pca[len(source_df):, 1], alpha=0.5, label="Target")
plt.title("PCA Projection of Source vs Target")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 重新生成脚本文件（因执行环境已重置）
import pandas as pd
from sklearn.ensemble import IsolationForest

# 读取数据文件（你可以修改这个路径）
input_path = "../data/target_all_data_b1.csv"
output_path = "../data/target_all_data_b1_cleaned.csv"


df = pd.read_csv(input_path)
feature_cols = [col for col in df.columns if col.lower() != "timestamp"]

# 为每列进行异常检测，并替换为邻近值均值
for col in feature_cols:
    model = IsolationForest(contamination=0.01, random_state=42)
    preds = model.fit_predict(df[[col]])  # -1 表示异常

    anomaly_indices = df[preds == -1].index
    for idx in anomaly_indices:
        # 取前后 2 个点的均值（不包括自己）
        neighbors = []
        if idx > 0:
            neighbors.append(df.at[idx - 1, col])
        if idx < len(df) - 1:
            neighbors.append(df.at[idx + 1, col])
        if neighbors:
            df.at[idx, col] = sum(neighbors) / len(neighbors)

df.to_csv(output_path, index=False)
print(f"处理完毕，保存为 {output_path}")
```
