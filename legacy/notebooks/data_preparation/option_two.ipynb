```python
import pandas as pd

# 读取 CSV 文件
df = pd.read_csv(r"F:\OneDrive - CentraleSupelec\PhD\Paper-Fault_detection\codes\models_reg_v3\experiments_source_v3\exp_20250411_141442_b1\predictions_results_n_past_30_n_future_10_20250411_141442.csv", sep=",", header=None)

# 假设前两列为数据集名（如 Test 或 Validation）和选项（如 Option 1 或 Option 2）
# 过滤包含 Test + Option 2 的行
df_test_option2 = df[(df[0] == "Test") & (df[1] == "Option 2")]

# 过滤包含 Validation + Option 2 的行
df_val_option2 = df[(df[0] == "Validation") & (df[1] == "Option 2")]

# 保存结果到新文件
df_test_option2.to_csv("filtered_test_option2.csv", index=False, header=False)
df_val_option2.to_csv("filtered_validation_option2.csv", index=False, header=False)

print("Done. Filtered rows saved to 'filtered_test_option2.csv' and 'filtered_validation_option2.csv'.")
```

```python
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error, r2_score
import numpy as np

# 读取 CSV 文件
df = pd.read_csv("filtered_validation_option2.csv", header=None)

# 假设最后两列为真实值和预测值
true_values = df.iloc[:, -2]
pred_values = df.iloc[:, -1]

# 计算指标
mse = mean_squared_error(true_values, pred_values)
rmse = np.sqrt(mse)
mape = mean_absolute_percentage_error(true_values, pred_values)
r2 = r2_score(true_values, pred_values)

# 输出结果
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"MAPE: {mape:.4%}")
print(f"R²:   {r2:.4f}")
```
